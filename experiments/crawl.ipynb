```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import json
```

```python
page  = f"https://www.topcv.vn/tim-viec-lam-software-engineering-cr257cb258?type_keyword=1&page=1&category_family=r257%7Eb258_b259_b260_b261_b262_b263&sba=1"

hrefs = []
for i in range(10):
    driver = webdriver.Chrome()
    driver.get(page)

    box_elements = driver.find_elements(By.XPATH, '//div[@data-box="BoxSearchResult"]')

    for box in box_elements:
        # Trong mỗi box, tìm div có class chứa 'avata' (hoặc 'avatar' nếu đúng)
        avatar_divs = box.find_elements(By.XPATH, './/div[contains(@class, "avatar")]')
        for avatar_div in avatar_divs:
            a_tags = avatar_div.find_elements(By.XPATH, './a')
            for a in a_tags:
                href = a.get_attribute('href')
                if href:
                    hrefs.append(href)
    driver.quit()
```

```python
try:
    next_button = driver.find_element(By.XPATH, '//a[@aria-label="Next »"]')
    next_button.click()
except Exception as e:
    print("Không tìm thấy nút Next hoặc không thể click:", e)
```

```python
with open("job_link.txt", "w") as f:
    f.write("\n".join(list(set(hrefs))))
    
```

```python
for i, link in enumerate(hrefs):
    driver = webdriver.Chrome()
    driver.get(link)
    try:
        info_div = driver.find_element(By.CLASS_NAME, 'job-detail__information-container')
        text = info_div.text

        with open(f'job/{i}.txt', 'w', encoding="utf-8") as f:
            f.write(text)
    except:
        print(i)
    driver.quit()
```
