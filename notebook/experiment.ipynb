````python
import json

def parse_experience(data_json:str):
    """
    Parse experience data from JSON list into list of dicts with skill and years.
    
    Args:
        data_json (str): JSON-formatted string of skills with years.
        
    Returns:
        List[Dict[str, Union[str, float]]]: Parsed list of skills and years.
    """
    data = json.loads(data_json.replace("```json", "").replace("```", ""))
    result = []

    for entry in data:
        if " - " in entry:
            try:
                skill, years_str = entry.split(" - ")
                years = float(years_str.split()[0])  # extract number before "year"/"years"
                result.append({"skill": skill.strip(), "years": years})
            except:
                print("error when processing in parse_experience, entry:", entry)
    
    return result
parse_experience(extract_infor)
````
