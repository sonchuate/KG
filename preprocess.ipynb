```python
from src.preprocess.utils import process_string
MAX_DISTANCE = 10
def get_sim(cv_graph:str, jd_graph:str):
    cv_entities_list, _ = entity_extractor.parse_entities(cv_graph)

    cv_node = [e.node() for e in cv_entities_list]
    for id, n in enumerate(cv_node):
        cv_node[id].label = 'h'
    new_cv_nodes = []
    for n in cv_node:
        new_cv_nodes.append(n)
        new_cv_nodes.extend([i.node for i in gm.get_relations_to_node(n)])

    new_cv_nodes = [f"{n.properties['name']}" for n in new_cv_nodes]
    new_cv_nodes = list(set(new_cv_nodes))
    print("\n".join(set(new_cv_nodes)))
    print('*'*80)

    jd_entities_list, _ = entity_extractor.parse_entities(jd_graph)
    jd_nodes = [e.node() for e in jd_entities_list]
    for id, n in enumerate(jd_nodes):
        jd_nodes[id].label = 'h'
    jd_nodes = [f"{n.properties['name']}" for n in jd_nodes]
    # print("\n".join(set(jd_node)))
    
    distance = []
    for cv_node in new_cv_nodes:
        tmp = []
        for jd_node in jd_nodes:
            d = gm.shortest_distance(cv_node, jd_node)
            if d == -1:
                tmp.append(MAX_DISTANCE)
            else:
                tmp.append(d)
        distance.append(min(tmp))
    return sum(distance)

    
    
```
